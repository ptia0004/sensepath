```python
import pandas as pd

path = "buildings.csv"
bldg = pd.read_csv(path)

bldg.head()
```

```python
print("Rows and columns:", bldg.shape)

print("\nColumn names:")
print(bldg.columns.tolist())

print("\nData types:")
print(bldg.dtypes)

print("\nFirst 5 rows:")
display(bldg.head())
```

```python
miss = pd.DataFrame({
    "missingcount": bldg.isna().sum(),
    "missingpercent": bldg.isna().mean() * 100
}).sort_values("missingpercent", ascending=False)

display(miss)
```

```python
dups = bldg.duplicated().sum()

print("Duplicate rows:", dups)
```

```python
desc = bldg.describe(include="all").T

display(desc)
```

```python
print(bldg["Census year"].min())
print(bldg["Census year"].max())

yr = bldg["Census year"].max()
bld = bldg[
    bldg["Census year"] == yr
].copy()

print("Latest year:", yr)
print("Rows:", bld.shape)
```

```python
drop_col = [
    "Has showers",
    "Refurbished year",
    "Bicycle spaces"
]

bld = bld.drop(
    columns=drop_col,
    errors="ignore"
)

print(bld.columns.tolist())
```

```python
bld = bld.dropna(
    subset=["Latitude", "Longitude"]
).copy()

print("Rows with coordinates:", len(bld))
```

```python
uses = bld["Predominant space use"].value_counts(
    dropna=False
)

display(uses)
```

```python
keep_use = [
    "Community Use",
    "Public Display Area",
    "Educational/Research"
]

cand = bld[
    bld["Predominant space use"].isin(keep_use)
].copy()

print("Candidate buildings:", len(cand))

display(
    cand[
        [
            "Building name",
            "Street address",
            "Predominant space use",
            "Accessibility type",
            "Accessibility rating",
            "Latitude",
            "Longitude"
        ]
    ].head(30)
)
```

```python
libr_b = bld[
    bld["Building name"].str.contains(
        "Library",
        case=False,
        na=False
    )
    |
    bld["Street address"].str.contains(
        "Library",
        case=False,
        na=False
    )
].copy()

print("Library buildings:", len(libr_b))

display(
    libr_b[
        [
            "Building name",
            "Street address",
            "Predominant space use",
            "Accessibility type",
            "Accessibility rating",
            "Latitude",
            "Longitude"
        ]
    ]
)
```

```python
bld_ref = pd.concat(
    [cand, libr_b],
    ignore_index=True
)

bld_ref = bld_ref.drop_duplicates(
    subset=["Property ID", "Latitude", "Longitude"]
)

print("Building candidates:", len(bld_ref))

display(
    bld_ref[
        [
            "Building name",
            "Street address",
            "Predominant space use",
            "Accessibility type",
            "Accessibility rating",
            "Latitude",
            "Longitude"
        ]
    ].head(30)
)
```

```python
bld_final = bld_ref[
    bld_ref["Predominant space use"].isin(
        ["Community Use", "Public Display Area"]
    )
    |
    bld_ref["Building name"].str.contains(
        "Library",
        case=False,
        na=False
    )
].copy()

print("Refined building candidates:", len(bld_final))

display(
    bld_final[
        [
            "Building name",
            "Street address",
            "Predominant space use",
            "Accessibility type",
            "Accessibility rating",
            "Latitude",
            "Longitude"
        ]
    ].head(50)
)
```

```python
print("Final candidate rows:", len(bld_final))

print("\nSpace use:")
print(
    bld_final["Predominant space use"].value_counts()
)

print("\nAccessibility:")
print(
    bld_final["Accessibility type"].value_counts(
        dropna=False
    )
)
```

```python
print(
    "Missing building names:",
    bld_final["Building name"].isna().sum()
)
```

```python
no_name = bld_final[
    bld_final["Building name"].isna()
].copy()

display(
    no_name[
        [
            "Street address",
            "Predominant space use",
            "Accessibility type",
            "Accessibility rating",
            "Latitude",
            "Longitude"
        ]
    ]
)
```

```python
review_b = bld_final[
    bld_final["Building name"].isna()
].copy()

bld_final = bld_final[
    bld_final["Building name"].notna()
].copy()

print("Need manual review:", len(review_b))
print("Named candidates:", len(bld_final))
```

```python
review_b.to_csv(
    "buildings_missing_names_review.csv",
    index=False
)
```

```python
odd_use = bld_final[
    bld_final["Predominant space use"].isin(
        [
            "Unoccupied - Under Construction",
            "Performances, Conferences, Ceremonies"
        ]
    )
]

display(
    odd_use[
        [
            "Building name",
            "Street address",
            "Predominant space use",
            "Accessibility type",
            "Accessibility rating"
        ]
    ]
)
```

```python
bld_final = bld_final[
    ~(
        bld_final["Building name"].eq(
            "State Library Rail Station"
        )
        &
        bld_final["Predominant space use"].eq(
            "Unoccupied - Under Construction"
        )
    )
].copy()

print("Rows after removal:", len(bld_final))
```

```python
print(
    bld_final[
        bld_final["Building name"].str.contains(
            "Athenaeum|State Library Rail",
            case=False,
            na=False
        )
    ][
        [
            "Building name",
            "Street address",
            "Predominant space use",
            "Accessibility rating"
        ]
    ]
)
```
