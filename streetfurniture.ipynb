```python
import pandas as pd

path = "street_furniture.csv"
furn = pd.read_csv(path)

furn.head()
```

```python
print("Rows and columns:", furn.shape)

print("\nColumn names:")
print(furn.columns.tolist())

print("\nData types:")
print(furn.dtypes)

print("\nFirst 5 rows:")
display(furn.head())
```

```python
miss = pd.DataFrame({
    "missingcount": furn.isna().sum(),
    "missingpercent": furn.isna().mean() * 100
}).sort_values("missingpercent", ascending=False)

display(miss)
```

```python
dups = furn.duplicated().sum()

print("Duplicate rows:", dups)
```

```python
types = furn["ASSET_TYPE"].value_counts(
    dropna=False
)

display(types)
```

```python
core = [
    "Seat",
    "Drinking Fountain",
    "Picnic Setting"
]

extra = [
    "Barbeque",
    "Information Pillar"
]
```

```python
extra_fac = furn[
    furn["ASSET_TYPE"].isin(extra)
].copy()

print("Extra facilities:", len(extra_fac))

print(
    extra_fac["ASSET_TYPE"].value_counts()
)
```

```python
cord = furn["CoordinateLocation"].str.split(
    ",",
    expand=True
)

furn["lat"] = pd.to_numeric(
    cord[0].str.strip(),
    errors="coerce"
)

furn["lon"] = pd.to_numeric(
    cord[1].str.strip(),
    errors="coerce"
)

display(
    furn[
        [
            "ASSET_TYPE",
            "DESCRIPTION",
            "CoordinateLocation",
            "lat",
            "lon"
        ]
    ].head()
)
```

```python
print("Missing lat:", furn["lat"].isna().sum())
print("Missing lon:", furn["lon"].isna().sum())

print("Latitude:", furn["lat"].min(), "to", furn["lat"].max())
print("Longitude:", furn["lon"].min(), "to", furn["lon"].max())
```

```python
core = [
    "Seat",
    "Drinking Fountain",
    "Picnic Setting"
]

extra = [
    "Barbeque",
    "Information Pillar"
]

keep = core + extra


furniture_clean = furn[
    furn["ASSET_TYPE"].isin(keep)
].copy()


print("Final furniture facilities:", len(furniture_clean))

print(
    furniture_clean["ASSET_TYPE"].value_counts()
)
```

```python
display(
    furniture_clean[
        [
        "GIS_ID",
        "ASSET_TYPE",
        "DESCRIPTION",
        "LOCATION_DESC",
        "CONDITION_RATING",
        "EVALUATION_DATE",
        "lat",
        "lon"
        ]
    ].head(30)
)
```

```python
print(
    furniture_clean["CONDITION_RATING"].describe()
)

print(
    furniture_clean["CONDITION_RATING"].isna().sum()
)
```

```python
lowc = furniture_clean[
    furniture_clean["CONDITION_RATING"].notna()
    &
    (furniture_clean["CONDITION_RATING"] < 2)
].copy()

print("Low condition facilities:", len(lowc))

display(
    lowc[
        [
            "ASSET_TYPE",
            "DESCRIPTION",
            "LOCATION_DESC",
            "CONDITION_RATING"
        ]
    ].head(20)
)
```

```python
def condition_category(x):
    if x >= 4:
        return "Good"
    elif x >= 2:
        return "Average"
    else:
        return "Poor"


furniture_clean["CONDITION_CATEGORY"] = (
    furniture_clean["CONDITION_RATING"]
    .apply(condition_category)
)


print(
    furniture_clean["CONDITION_CATEGORY"].value_counts()
)
```
