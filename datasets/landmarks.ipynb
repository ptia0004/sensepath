```python
import pandas as pd

path = "landmarks.csv"
data = pd.read_csv(path)

data.head()
```

```python
print("Rows and columns:", data.shape)

print("Column names:")
print(data.columns.tolist())

print("Data types:")
print(data.dtypes)

print("First 5 rows:")
display(data.head())
```

```python
miss = data.isna().sum().sort_values(ascending=False)

display(miss)
```

```python
miss_sum = pd.DataFrame({
    "missingcount": data.isna().sum(),
    "missingpercent": data.isna().mean() * 100
}).sort_values("missingpercent", ascending=False)

display(miss_sum)
```

```python
dups = data.duplicated().sum()

print("Duplicate rows:", dups)
```

```python
for col in ["Theme", "Sub Theme"]:
    if col in data.columns:
        print(f"\nUnique values in {col}:")
        print(data[col].value_counts(dropna=False).head(30))
```

```python
desc = data.describe(include="all").T

display(desc)
```

```python
geo_col = "Co-ordinates"

print(data[geo_col].head())
print("Missing coordinates:", data[geo_col].isna().sum())
print("Unique coordinates:", data[geo_col].nunique())
```

```python
cord = data["Co-ordinates"].str.split(",", expand=True)

data["lat"] = pd.to_numeric(cord[0].str.strip(), errors="coerce")
data["lon"] = pd.to_numeric(cord[1].str.strip(), errors="coerce")

display(data[["Feature Name", "Co-ordinates", "lat", "lon"]].head())
```

```python
print("Missing latitude:", data["lat"].isna().sum())
print("Missing longitude:", data["lon"].isna().sum())
```

```python
print("Latitude:", data["lat"].min(), "to", data["lat"].max())
print("Longitude:", data["lon"].min(), "to", data["lon"].max())
```

```python
name_cnt = data["Feature Name"].value_counts()

name_dup = name_cnt[name_cnt > 1]
display(name_dup)
```

```python
same = data[data["Feature Name"].isin(name_dup.index)]

display(
    same.sort_values("Feature Name")
)
```

```python
park = data[
    data["Sub Theme"].eq(
        "Informal Outdoor Facility (Park/Garden/Reserve)"
    )
].copy()

print("Park records:", len(park))
display(park.head(10))
```

```python
libr = data[
    data["Feature Name"].str.contains(
        "Library",
        case=False,
        na=False
    )
    |
    data["Sub Theme"].str.contains(
        "Library",
        case=False,
        na=False
    )
].copy()

keep = [
    "Public Buildings",
    "Visitor Centre",
    "Art Gallery/Museum"
]

publ = data[
    data["Sub Theme"].isin(keep)
].copy()

print("Parks:", len(park))
print("Libraries:", len(libr))
print("Candidate public places:", len(publ))
```

```python
refu = pd.concat(
    [park, libr, publ],
    ignore_index=True
)

refu = refu.drop_duplicates(
    subset=["Feature Name", "lat", "lon"]
)

print("Total candidates:", len(refu))
display(refu.head(20))
```

```python
print("Dataset size:", data.shape)
print("Missing values:", int(data.isna().sum().sum()))
print("Duplicate rows:", int(data.duplicated().sum()))
print("Unique themes:", data["Theme"].nunique())
print("Unique sub themes:", data["Sub Theme"].nunique())
print("Unique feature names:", data["Feature Name"].nunique())
print("Unique coordinates:", data["Co-ordinates"].nunique())
print("Park records:", len(park))
print("Library records:", len(libr))
print("Possible public spaces:", len(publ))
```

```python
refu.to_csv(
    "landmarks_refuge_candidates.csv",
    index=False
)

print("Saved:", len(refu), "records")
```
